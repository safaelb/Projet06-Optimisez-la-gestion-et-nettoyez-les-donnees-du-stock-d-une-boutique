```python
#Fusion des fichiers df_erp et df_liaison
# S'assurer que la colonne de jointure est bien du même type
data_erp['product_id'] = data_erp['product_id'].astype(str)
data_liaison['product_id'] = data_liaison['product_id'].astype(str)

# Fusion en left pour garder tous les articles de l’ERP
data_fusionne = pd.merge(
    data_erp,
    data_liaison,
    how='left',
    on='product_id',
    indicator=True
)

print("Fusion ERP + Liaison effectuée.")
print("Dimensions :", data_fusionne.shape)
```

<div style="border: 1px solid RGB(51,165,182);" >
<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 3.2 - Jonction du fichier df_merge et df_web</h3>
</div>

```python
#Fusionnez les datasets df_merge et df_web
data_fusionne['id_web'] = data_fusionne['id_web'].astype(str)
data_web['sku'] = data_web['sku'].astype(str)

# Fusion
data_web_fusionne = pd.merge(
    data_fusionne,
    data_web,
    how='outer',
    left_on='id_web',
    right_on='sku',
    suffixes=('', '_web')
)
print("Fusion réussie")
print(data_web_fusionne.shape)
```

```python
#Avons-nous des lignes sans correspondances?

lignes_sans_correspondance = data_web_fusionne[data_web_fusionne['sku'].isnull()]
print("\ndes lignes sans correspondances :")
print(lignes_sans_correspondance)

data_web_fusionne = data_web_fusionne[~data_web_fusionne['sku'].isnull()]
print(f"Fusion réussie : {data_web_fusionne.shape[0]} lignes avec correspondance.")
```

```python
#Calculer la moyenne du prix
moyenne_prix = data_web_fusionne["price"].mean()
print("Moyenne du prix :", moyenne_prix)

#Calculer l'écart-type du prix
ecart_type_prix = data_web_fusionne["price"].std()
print("Écart-type du prix :", ecart_type_prix)
#Calculer le Z-score
data_web_fusionne["z_score"] =  moyenne_prix / ecart_type_prix

# Afficher un aperçu des Z-scores
print(data_web_fusionne[["product_id", "price", "z_score"]].head())

```

```python
#Quel est le seuil prix dont z-score est supérieur à 3?
seuil_prix = 3 * ecart_type_prix + moyenne_prix
print(f"Le seuil de prix où Z-score > 3 est : {seuil_prix}")
```

<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 4.2.2 - Identification par l'interval interquartile</h3>
</div>

```python
#Définissez un seuil pour les articles "outliers" en prix
# Calcul des quartiles
Q1 = data_web_fusionne['price'].quantile(0.25)  # Premier quartile (25%)
Q3 = data_web_fusionne['price'].quantile(0.75)  # Troisième quartile (75%)
IQR = Q3 - Q1  # Intervalle interquartile

# Calcul des seuils
seuil_inferieur = Q1 - 1.5 * IQR  # Seuil inférieur
seuil_superieur = Q3 + 1.5 * IQR  # Seuil supérieur

# Identification des outliers
outliers = data_web_fusionne[(data_web_fusionne['price'] < seuil_inferieur) | (data_web_fusionne['price'] > seuil_superieur)]

# Affichage des résultats
print(f"Seuil inférieur : {seuil_inferieur}")
print(f"Seuil supérieur : {seuil_superieur}")
print(f"Nombre d'outliers : {len(outliers)}")
print(f"Proportion d'outliers : {len(outliers) / len(data_erp):.2%}")

```

```python
#############################
# Calculer le 20 / 80 en CA #
#############################
# 1. On garde les lignes valides avec price et total_sales non null
df_valide = data_web_fusionne.dropna(subset=['price', 'total_sales']).copy()

# 2. Calcul du CA par article
df_valide['ca_par_article'] = df_valide['price'] * df_valide['total_sales']

# 3. Tri décroissant du CA
df_valide = df_valide.sort_values(by='ca_par_article', ascending=False)

# 4. Calcul de la part du CA de chaque ligne
ca_total = df_valide['ca_par_article'].sum()
df_valide['part_ca'] = df_valide['ca_par_article'] / ca_total

# 5. Calcul de la somme cumulative
df_valide['ca_cumulée'] = df_valide['part_ca'].cumsum()

# 6. Identifier combien d’articles représentent 80 % du CA
seuil_80 = df_valide[df_valide['ca_cumulée'] <= 0.80]
nb_articles_80 = seuil_80.shape[0]
print(f"Nombre d'articles représentant 80 % du CA : {nb_articles_80}")

# 7. Calculer la proportion dans le catalogue
total_articles = df_valide.shape[0]
proportion = nb_articles_80 / total_articles * 100
print(f"Ils représentent {proportion:.2f} % du catalogue (total articles : {total_articles})")
```

```python
##############################################
# Analyse du taux de marge #
##############################################

# Création de la colonne prix HT
# On suppose que le prix TTC est dans la colonne "price" et que la TVA est de 20%
taux_tva = 0.20
data_web_fusionne['prix_ht'] = data_web_fusionne['price'] / (1 + taux_tva)
data_web_fusionne['purchase_price_ht'] = data_web_fusionne['purchase_price'] / (1 + taux_tva)


# Création de la colonne Taux de marge
# Le taux de marge est calculé comme suit : (Prix de vente - Coût d'achat) / Prix de vente
# On suppose que le coût d'achat est dans la colonne "purchase_price"
data_web_fusionne['taux_marge'] = (data_web_fusionne['price'] - data_web_fusionne['purchase_price']) / data_web_fusionne['price']

# Afficher le prix minimum de la colonne "taux_marge"
min_taux_marge = data_web_fusionne['taux_marge'].min()
print(f"Le taux de marge minimum est : {min_taux_marge:.2f}")

# Afficher le prix maximum de la colonne "taux_marge"
max_taux_marge = data_web_fusionne['taux_marge'].max()
print(f"Le taux de marge maximum est : {max_taux_marge:.2f}")
```
